```
# IPL Data Analysis Using Python Pandas

import pandas as pd
import numpy as np

# Load Dataset with a fallback mock creation if file doesn't exist
try:
    df = pd.read_csv("IPL.csv")
    print("Loaded data from IPL.csv\n")
except FileNotFoundError:
    print("IPL.csv not found. Generating mock IPL data for analysis...\n")
    # Generate simulated IPL analytical dataset
    np.random.seed(42)
    teams = ["Mumbai Indians", "Chennai Super Kings", "Royal Challengers Bangalore", "Kolkata Knight Riders", "Delhi Capitals"]
    batsmen_list = ["Virat Kohli", "Rohit Sharma", "MS Dhoni", "Suresh Raina", "Shikhar Dhawan", "AB de Villiers"]
    bowlers_list = ["Jasprit Bumrah", "Yuzvendra Chahal", "Lasith Malinga", "Amit Mishra", "Ravichandran Ashwin"]

    mock_data = []
    for match in range(1, 101):
        season = np.random.choice([2019, 2020, 2021, 2022])
        bat_team = np.random.choice(teams)
        bowl_team = np.random.choice([t for t in teams if t != bat_team])
        winner = np.random.choice([bat_team, bowl_team])

        # Generate a few delivery/over aggregates per match
        for _ in range(15):
            mock_data.append({
                "match_id": match,
                "season": season,
                "batting_team": bat_team,
                "match_won_by": winner,
                "batter": np.random.choice(batsmen_list),
                "runs_batter": np.random.choice([0, 1, 2, 4, 6], p=[0.3, 0.4, 0.1, 0.12, 0.08]),
                "bowler": np.random.choice(bowlers_list),
                "bowler_wicket": np.random.choice([0, 1], p=[0.9, 0.1]),
                "runs_total": np.random.randint(1, 8)
            })
    df = pd.DataFrame(mock_data)

# Display first 5 records
print("IPL Dataset:")
print(df.head())

# 1. Dataset Information
print("\nDataset Information:")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

# 2. Top 10 Run Scorers
print("\nTop 10 Batsmen:")
batsmen = df.groupby("batter")["runs_batter"].sum()
print(batsmen.sort_values(ascending=False).head(10))

# 3. Top 10 Wicket Takers
print("\nTop 10 Bowlers:")
bowlers = df.groupby("bowler")["bowler_wicket"].sum()
print(bowlers.sort_values(ascending=False).head(10))

# 4. Total Runs by Teams
print("\nTeam Performance:")
teams = df.groupby("batting_team")["runs_total"].sum()
print(teams.sort_values(ascending=False))

# 5. Total Matches Per Season
print("\nMatches Per Season:")
matches = df.groupby("season")["match_id"].nunique()
print(matches)

# 6. Most Successful Teams
print("\nMatch Winning Teams:")
winners = df.groupby("match_won_by")["match_id"].nunique()
print(winners.sort_values(ascending=False).head(5))

# 7. Highest Individual Score
print("\nHighest Individual Score:")
scores = df.groupby(["match_id", "batter"])["runs_batter"].sum()
print(scores.sort_values(ascending=False).head(5))

# Final Result
print("\nIPL DATA ANALYSIS COMPLETED!")
```
